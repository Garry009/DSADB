```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('custom_data.csv')
pt1=df['reading_score'].head(25)
pt2=df['math_score'].head(25)
```

```python
plt.plot(pt1,color="g", marker='X',mec='b', ms=5,mfc='r')
plt.plot(pt2,color="r", marker='s',mec='y', ms=10,mfc='C9')
plt.xlabel("Rollno")
plt.ylabel("Marks")
plt.legend(['Reading Score','Math Score'])
plt.plot(pt2)
plt.axhline(y=df['math_score'].mean(), color='r', linestyle='--', label=f'Mean: {df['math_score'].mean()}')
plt.axhline(y=df['reading_score'].mean(), color='r', linestyle='--', label=f'Mean: {df['reading_score'].mean()}')
plt.savefig("fig1.pdf")
plt.show()
```
